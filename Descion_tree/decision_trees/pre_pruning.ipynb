## Post pruning

```python
import pandas as pd 
import matplotlib.pyplot as plt 
```

```python
from sklearn.datasets import load_iris
```

```python
iris=load_iris()
```

```python
import seaborn as sns
df=sns.load_dataset('iris')
```

```python
x=df.iloc[:,:-1]
y=iris.target
```

```python
x,y
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.33,random_state=42)
```

```python
x_train
```

### post pruning

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
parameter= {'criterion':['gini','entropy','log_loss'],'splitter':['best','random'],'max_depth':[1,2,3,4,5],'max_features':['auto','sqrt','log2']}
```

```python
treemodel = DecisionTreeClassifier()
gridcv=GridSearchCV(treemodel,param_grid=parameter,cv=5,scoring='accuracy')
```

```python
gridcv.fit(x_train,y_train)
```

```python
gridcv.best_params_
```

```python
from sklearn.metrics import accuracy_score
tpred = gridcv.predict(x_test)
testscore= accuracy_score(tpred,y_test)
testscore
```
