```python
import pandas as pd
data=pd.read_csv('datasets\Life Expectancy Data.csv')
```

```python
data.head()
```

```python
data.isnull().sum()
```

```python
x=data.drop(['Life expectancy ','Status','Year','Country'],axis=1)
```

```python
x.head()
```

```python
y=data[["Life expectancy "]]
```

```python
from sklearn.impute import SimpleImputer
strat_mean=SimpleImputer(strategy='mean')
```

```python
x=strat_mean.fit_transform(x)
y=strat_mean.fit_transform(y)
```

```python
X=pd.DataFrame(x)
X
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(x,y,random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
X_train_scale=pd.DataFrame(scaler.fit_transform(X_train))
X_test_scale=pd.DataFrame(scaler.transform(X_test))
```

```python
X_test_scale
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
```

```python
model.fit(X_train_scale,Y_train)
```

```python
yhat=model.predict(X_train_scale)
```

```python
y_predict=model.predict(X_test_scale)
```

```python
model.score(X_test_scale,Y_test)
```

```python
from sklearn.metrics import mean_squared_error
mse=(mean_squared_error(Y_train,yhat)/2)
mse2=(mean_squared_error(Y_test,y_predict)/2)
print(mse,mse2)
```
