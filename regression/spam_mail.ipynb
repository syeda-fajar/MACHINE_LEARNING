```python
import numpy as np 
import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score,confusion_matrix
```

```python
raw_mail_data = pd.read_csv('datasets/mail_data.csv')
```

```python
raw_mail_data.head()
```

## Handling missing value

```python
# replace NUll vaules with null string
mail_data = raw_mail_data.where((pd.notnull(raw_mail_data)),'')
```

```python
mail_data.head()
```

```python
mail_data.shape
```

```python
mail_data['Category']=mail_data['Category'].map({'ham':1,'spam':0})
```

```python
mail_data.head()
```

```python
x = mail_data['Message']
y= mail_data['Category']
```

```python
x.head()
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=3)
```

```python
print(x_train.shape,x_test.shape,y_train.shape,y_test.shape)
```

### Feature Extraction

- transforming text data into feature vector that can be used as input to logistic regression

```python
feature_extraction =TfidfVectorizer(min_df=1,stop_words='english',lowercase=True)
x_train_feature = feature_extraction.fit_transform(x_train)
x_test_feature = feature_extraction.transform(x_test)
```

```python
print(x_train_feature)
```

```python
model=LogisticRegression()
model.fit(x_train_feature,y_train)
pred = model.predict(x_train_feature)
score = accuracy_score(pred,y_train)
score
```

```python
pred_test = model.predict(x_test_feature)
score_test = accuracy_score(pred_test,y_test)
score_test
```
